```python
#HOF Usecase1:
# Math Calculator: Write a calculate(a, b, operation) function. Pass small helper functions like add() or multiply() directly into the operation argument instead of using messy if/else blocks.
def add(a,b):
    return a+b

def sub(a,b):
    return a-b

def mul(a,b):
    return a*b

def div(a,b):
    return a/b

def wrapper_hof1_fun(a,b,fun):#We can pass value or function as an input to a function
    #print(a,b,xyz)#a=10,b=5,xyz=add
    result=fun(a,b)
    return result

print(wrapper_hof1_fun(10,5,add))
print(wrapper_hof1_fun(10,5,sub))
print(wrapper_hof1_fun(10,5,mul))
print(wrapper_hof1_fun(10,5,div))
```

```python
#HOF Type1
#Create a function to Implement AC remote operation
#ICs/Electric/Electronic modules
def cool(temp):
    print(f"cool temp is set to {temp}")
    return temp
def dry(temp):
    print(f"dry temp is set to {temp}")
    return temp

def clean():
    try:
        import time
        print("cleaning started")
        time.sleep(5)
        print("cleaning completed")
        return True
    except Exception as e:
        print("cleaning failed")
        return False

def hitachi_ac_hof(mode,value=None):
    #Here i will write all other functionalities like compressor, cooling gas
    #I don't want to write cool,dry,clean logic rather i will take it as an argument.
    if mode ==clean:
        return mode()
    else:
        return mode(value)
hitachi_ac_hof(cool,24)#eg. of calling a function with value & code
hitachi_ac_hof(value=30,mode=dry)#eg. of calling a function with value & code
print(hitachi_ac_hof(clean))

def xyz_ac_hof(mode,value):
    mode(value)
xyz_ac_hof(cool,24)
xyz_ac_hof(value=30,mode=dry)
#We don't have clean option in this AC
```

```python
#HOF Type2 - A Function Returns another function
#Create a function to Implement AC remote mode changing operation
#ICs/Electric/Electronic modules
def ac_remote(mode):
    if mode==1:
        def cool(temp):
            print(f"cool temp is set to {temp}")
            return temp
        return cool
    elif mode==2:
        def dry(temp):
            print(f"dry temp is set to {temp}")
            return temp
        return dry
    else:
        def clean():
            try:
                import time
                print("cleaning started")
                time.sleep(5)
                print("cleaning completed")
                return True
            except Exception as e:
                print("cleaning failed")
                return False
        return clean

ac_remote(1)(25)
#or
parent_ret_child=ac_remote(1)#Calling parent that is returning child function, storing function code in a variable
parent_ret_child(25)# then calling childfunction(variable) with temp as arg, that returns result
#print(parent_fun_call_returns_child(25))
ac_remote(2)(30)#Calling parent that is returning child then calling child(with arg) that returns result
#print(ac_remote(2))
ac_remote(3)()#Calling parent that is returning child then calling child(without arg) that returns result
```

```python
#Rules to create Lambda functions.
#Lambda function has to anonymous function within the module/notebook scope, not outside of it.
#Eg: If I need SOME SIMPLE + QUICK functionality to use multiple times WITHIN this notebook, then i will create lambda function, because lambda function has limitation of not supporting block of code, it supports only a line of code in the body
calc_bonus=lambda sal,bon_pct : sal+(sal*bon_pct)
```

```python
#Lambda Function Usecase:
# Daily High Temperatures: Suppose you have a list of daily temperature readings grouped by hours. You want to quickly extract the highest temperature for each day. You can use a one-liner
temp_lst=[10,15,20,30,35,30,20]
irfans_hand=[10,15,20,30,35,30,40,30]
def lam_stats_fun(saravanans_hand):
    return sum(saravanans_hand)/len(saravanans_hand)

result_def_tuple=lam_stats_fun(saravanans_hand=temp_lst)
result_def_tuple=lam_stats_fun(saravanans_hand=irfans_hand)

print("Avg Temp",result_def_tuple)

#both are same
var_lam_stats_fun=lambda vijay : (sum(vijay)/len(vijay), min(vijay), max(vijay))
result_tuple=var_lam_stats_fun(temp_lst)
print("Avg Temp",result_tuple[0],"min temp",result_tuple[1],"max temp",result_tuple[2])
```

```python
#Closure: If a variable defined outside of the function that impacts the function result.
store_wide_offer=.08#global var
def purchase(amt):
    return amt-(amt*store_wide_offer)#If the global var is going to impact the result of a function (ie. CLOSURE)
purchase(10000)
```
