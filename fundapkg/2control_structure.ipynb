```python
#min_cart_amt,disc_pct_whole, max_disc_amt = 600,10,100
min_cart_amt = 600
disc_pct_whole = 10
max_disc_amt = 100
#print(min_cart_amt, disc_pct_whole, max_disc_amt)
#cart_value=float(input("enter cart amount"))
cart_value = 1200
if cart_value>=min_cart_amt:#True - lines 6,7,8,9,10, False - line 17
    print("U r eligible for offer")
    disc_pct=disc_pct_whole/100
    #disc_applied_cart_amt = - cart_value * disc_pct #unary operator
    disc_applied_cart_amt =  cart_value * disc_pct #Assignment  & Binary too
    print(f"disc_pct {disc_pct}% disc_pct_applied_cart_amt {disc_applied_cart_amt}")
    #print(abs(disc_applied_cart_amt))#abs will return positive value always
    if disc_applied_cart_amt<=max_disc_amt:#True - lines 11, 12 #False - lines 14, 15
        final_bill_amt=cart_value - disc_applied_cart_amt
        print(f"final bill amount is after applying max discount amount of {disc_applied_cart_amt} on {cart_value} is {final_bill_amt}")
    else:
        final_bill_amt=cart_value-max_disc_amt
        print(f"final bill amount is after applying max discount amount of {max_disc_amt} on {cart_value} is {final_bill_amt}")
else:
    print(f"Add some more amout to get offer {min_cart_amt-cart_value}")
```

Looping Construct

**Looping**
Iterative or repetitive execution of similar tasks across different data or programs (with different input or/and until condition(s) are fulfilled).
**Unconditional - Repetitive/Recursive/Iterative/Looped execution of the same logic with different values**
**Conditional - Repeating a set of statements again and again until a condition is satisfied**

```python
#Only 2 types of loops in python
#for loop & while loop (do while)
#Unconditional Looping
#A loop where iterations are not controlled by a condition; it iterates over a predefined sequence.
#for loop is an unconditional looping
#Runs on top of all sequence/iterable types (collection types, range, string)
for i in [1,2,3]:
    print(i)

for i in (1,2,3):
    print(i)

for i in {1:10,2:20,3:30}.items():
    print(i)

for i in 'Inceptez':
    print(i)

import time
for i in range(1,10,2):#i=1 then print(i), i=3 then print(i)....
    print(type(i))
    print(i)
    time.sleep(2)
```

```python
#Implement Password validation. Keep prompting the user for a password indefinitely (simulating a do-while loop) and use a break statement to exit the loop only when the password meets all required conditions.

#What is entry controlled loop -
# If we control the execution of the loop at the time of entry itself
startval=1
endval=5
while startval<=endval:#entry controlled
    print(f"execute the loop {startval}")
    startval+=1

#What is exit controlled loop -
# If we control the execution of the loop at the time of exit
startval=1
endval=5
while True:#dont put any control in the entry, just allow the loop to run
    if startval>endval:#exit controlled loop
        print(f"terminate the loop because {startval} > {endval}")
        break #exit controlled loop
    print(f"execute the loop {startval}")
    startval+=1


print("after break, this line of code executed")
```

```python
#Usecase4: Implement Password validation. Keep prompting the user for a password infinately (simulating a do-while loop) and use a break statement to exit the loop only when the password meets all required conditions.
actual_password='Inceptez@123'
#In python we don't have exclusively do while looping, but we can use while True
# (exit control loop to achieve do while loop)
while True:#not entry control loop
    user_password=input("enter the password")
    if user_password==actual_password:
        print("you are allowed to access the application")
        break # Exit control
    else:
        print("retry the password")
```
