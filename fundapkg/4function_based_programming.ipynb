```python
#What is a function?
#Verbally : An object/Item does some functionality is a function.
#Verbal Example for a function:
#Eg. AC : Does some functionality (maintain temperation)
#Take some input, Produce some Output,
# Only it can take an input but no output is given,
# Produce output alone without an input given
#Programatically/Technically:
# A Program (one or more loc) that perform some functionality.
#Python is a FBP Language
```

```python
#reuse an existing function, rather than reinventing
import builtins
print(builtins.len([1,2,3,10,20]))#Reusability, Simple
print(len(list((1,2,3,10,20))))#Composability
```

```python
#Rather than reinvent, reuse
def irfan_len_func(lst):
    a=0
    for i in lst:
        a+=1
    return a
print(irfan_len_func([1,2,3,10,20]))
```
