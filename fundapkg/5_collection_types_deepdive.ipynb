```python
#Builtin Functions to access/updating/resizing
#How to select/Access element from list?
print(lst1[0])#using index

print("using map function we can select items")
print(list(map(lambda i:print(i+1000),lst1)))#select

print("regular for loop we can select items")
for i in lst1:#select
    print(i+1000)

##Access - Steps to access & understand a builtin function?
#0. Use AI
#1. extend our value with .
#2. Use the NAME to understand about the function basically
#3. HOVER the function to understand about it additionally using doc string
#4. Using documentations by the creater of the functions
print("before reverse",lst1)
lst1.reverse()#I am updating/mutating the given list
print("after reverse",lst1)#I am selecting the list by reversing

#Access - Filtering & selecting the count of a given value in a list
lst1.count(5000)#count is a method rather than calling it a function

#Access -
lst1=[10000,15000,25000,5000,12000,5000]
print(lst1.index(5000))
#position 1  2      4   4       5   6
#index 0    1       2   3       4   5
lst1=[10000,15000,25000,5000,12000,5000]
print(lst1.index(5000))#return 3
print(lst1.index(5000,4,6))
#I don't need the first index value of a input value?
print(lst1.index(5000,lst1.index(5000)+1))
print("Access from right to left using -1",lst1.index(5000,-1))
#lst1.reverse()
print(list(reversed(lst1)).index(5000))
print("reversed list in memory",list(reversed(lst1)))
print("original list is not affected",lst1)
```

```python
#Updatable/Mutable? yes
lst=[10000,15000,25000,5000,12000,5000]
print("original list",lst)
lst[1]=50000
print("updated/mutated",lst)
lst.reverse()#Update/mutate the list with reversed values
print("updated after reverse()",lst)
lst_copy=[1000]
lst_copy=lst.copy()#update function
print("list with 1000 got updated with lst values",lst_copy)
lst.sort(reverse=True)
print("updated after sort() (mutable function) in a descending order",lst)
```

```python
#Resizing functions (insert/delete)? yes
lst=[10000,15000,25000,5000,12000,5000]
print(len(lst))
lst.append(50000)#insert
print(lst)
print(len(lst))
tup1=(1000,2000,3000)
lst.extend(tup1)#insert any other seq/iterable types to our list
print(lst)
print(len(lst))
lst.pop(1)#delete last index element/given index value
print(len(lst))
print(lst)
lst.remove(5000)#delete the given value
print(len(lst))
print(lst)
lst.clear()#wipe off the list
print(len(lst))
print("clean list",lst)
```

```python
#dict
dic={1:'irfan',2:'uvaraj',3:'sojiv',1:'senthil'}
#select(access)
print(dic[2])#access using key returns value, if key is not present throws error keyerror
print(dic.get(2))#access using key returns value, if key is not present returns default of None
print(list(dic.items()))#returns both key & value in what format?
for key,value in dic.items():
    print(key,value)
print(dic.keys())#returns only keys in a format of list
print(dic.values())#returns only keys in a format of list

#updatable(mutable)
dic1={1:'irfan',2:'uvaraj',3:'sojiv'}
dic2={1:'senthil',4:'suji'}
dic1.update(dic2)#merge (update if same key else append)
print(dic1)
dic1[2]='Yuvaraj'
print(dic1)

#insert/delete(resizable)
dic1.pop(3)#pop will remove the item of the given key, keyerror will be thrown otherwise
print(dic1)
dic1={1:'irfan',2:'uvaraj',3:'sojiv'}
dic1.popitem()#last item of the dictionary
print(dic1)
dic1.update({4:'suji'})#Inserting/updating items to dictionary
print(dic1)
dic1.clear()
print(dic1)
dic1={1:'irfan',2:'uvaraj',3:'sojiv'}
dic3={}.fromkeys([1,2,3,4],"inceptez")#Generate a dictionary {1:'inceptez',2:'inceptez'}
print(dic3)
dic1={1:'irfan',2:'uvaraj',3:'sojiv'}
dic1.setdefault(4,"inceptez")#insert of the given dictionary using the key & value
print(dic1)#{1:'irfan',2:'uvaraj',3:'sojiv',4:'inceptez'}
```
