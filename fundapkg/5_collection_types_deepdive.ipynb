```python
#Collections/Complex
#Howmany:4
#order of importance: list, dict, tuple, set
#list[]: Indexed Seq/iterable collection of homogeneous elements/items
#dict{k:v..}: Seq/iterable collection of key,value pairs
#tuple(): Indexed Seq/iterable collection of hetrogeneous elements/items
#set{1,2..}: Seq/iterable collection of unique homogeneous elements/items
#All collection are Sequenced/iterable (run loop on all collection types)
#    0,1,2,3
lst=[1,2,3,4]
print(type(lst))
print(lst[0])
#      key
dict1={1:10,2:20,3:30,4:40}
#    0, 1,  2,  3  ,4
tup=(1,'a','b','c',100)
#    set operations/iterations
set1={1,2,3,3,2}

def iterate(coll_type):
    for elem in coll_type:
        print(coll_type)
        print(elem)
iterate(lst)
iterate(dict1)
iterate(tup)
iterate(set1)
```
