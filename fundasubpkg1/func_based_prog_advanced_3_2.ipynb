**Higher Order Function**

```python
#HOF Type1: A function that take another function as an input argument
#basically a function can be passed as input to another function
def fun1():
    return 'inceptez'

def fun2(arg):#arg=fun1 - arg is going to assigned with fun1 code
    print(arg)

fun2(fun1)#arg=fun1 code - call by reference
fun2(fun1())#arg='inceptez' - call by value (Composable)


```

```python
#HOF Type2: A function that can return another function as an output
def parent():
    print("parent is running with some logic & returning child")
    def child(a):
        print(f"child is running now with a param of {a}")
    return child
childvar=parent()#run parent & returns child
childvar(100)
```

```python
#HOF Type1 : A Function take another function as an argument
def bonus_calc(sal):
    return sal*.10
def inc_calc(sal):
    return 3000

def total_sal(sal,fun):
    bonus_inc_sal=fun(sal)#HOF type1
    def final_sal(sal,bonus_inc_sal):
        return sal+bonus_inc_sal
    return bonus_inc_sal,final_sal
#print(total_sal(10000,bonus_calc))#HOF type1
fun_returned=total_sal(20000,inc_calc)#HOF type1
fun_returned[1](20000,fun_returned[0])#HOF type1 & type2 (nested func/currying/partial func)
```

```python
#HOF - 1/2
#Type1 - fun(fun)
def cool(temp):
    print(f"logic to set cool temp of {temp}")
    return temp
def dry(intencity):
    print(f"logic to set dry intencity of {intencity}")
    return intencity
def clean():
        import time
        print(f"logic to do clean")
        time.sleep(10)
        return True

#XYZ brand which has only cool option
def ac_remote_xyz(value,fun):
    fun(value)
    print("temp set for xyz ac")

#Hitachi
def ac_remote_hitachi(*args):
    if args[0]==clean:
        mode=args[0]#clean()
        print('cleaning is ','complete' if mode()==True else 'failed')#clean()
    else:
        value=args[0]
        mode=args[1]
        print(f'for the mode {mode} value is set to {value}',mode(value))#cool(24) or dry(3)

ac_remote_xyz(24,cool)
ac_remote_hitachi(clean)#HOF type1 #Call by only reference
ac_remote_hitachi(24,cool)#Call by value & reference
ac_remote_hitachi(3,dry)
```

```python
#HOF Type2: A function that can return another function as an output
#Reallife Example: GPay payment -> 2 levels of protection - level1: entering into gpay app, level2: entering pin
actual_fingerprint='xyz'#closure
def entering_gpay(input_fingerprint):
    if input_fingerprint==actual_fingerprint:#Closure affects the function result
        print(f"You are entered into gpay using your fingerprint {input_fingerprint} successfully")
        print("I will return you the payment function enter_pin()")
        actual_pin='1234'
        def enter_pin_payment(pin,payment):
            if actual_pin==pin:
                print(f"Pin entered {pin} matches , hence allowing you to make payment {payment}")
        return enter_pin_payment

returned_enter_pin_payment_fun=entering_gpay('xyz')#Partial function/currying(partially cooked)/nested function/hof type2
returned_enter_pin_payment_fun('1234',1000)#equivalent to calling enter_pin_payment() function
#childvar(100)
```

```python
#Nested Function
def briyani():
    print("returning plain briyani")
    def salt():
        print("adding salt")
        def chilli():
            print("adding chilli")
            return 'ready to consume briyani'
        return chilli
    return salt

plainbriyani=briyani()#Currying
salt_added=plainbriyani()
chilli_added=salt_added()
print(chilli_added)
```
