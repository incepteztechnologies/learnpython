```python
min_cart_amt=600
disc_pct_whole=10
max_disc_amt=100
cart_amt=int(input("Choose the products hence amount added in the cart"))
if cart_amt>=min_cart_amt:
    print("eligible for offer")
    disc_pct_fraction=disc_pct_whole/100 # eg. 10/100 -> .10
    calc_discount_amt=cart_amt * disc_pct_fraction #eg. 800 *.10 -> 80
    if max_disc_amt >= calc_discount_amt:
        final_bill_amount=cart_amt-calc_discount_amt
        print(f"final bill amt after reducing discount of 10%: {calc_discount_amt} final bill amt {final_bill_amount}")
    else:
        final_bill_amount=cart_amt-max_disc_amt
        print(f"final bill amt after reducing discount flat discount of max_disc_amt :{max_disc_amt} final billl amt {final_bill_amount}")
else:
    print(f"not eligible for offer, add this amount in your cart {min_cart_amt-cart_amt}")
```

```python
#Looping Construct:
#Repetitive/Iterative/Recursive execution of a program based on some condition/values
#Types of Looping in any prog languages: 3 types (for, while, do while)
#Types of Looping in Python: 2 types (for, while)
#Categories of Looping: 2 categories (Unconditional(for is important in DE/AI) , Conditional(while is important in general))
#Unconditional eg. select * from tablename; -> for loop behind the scene to fetch all 10k recs
#Conditional eg. select * from tablename limit 10; -> while loop behind the scene to fetch only 10 records
```

```python
#Lets understand fundamentally - What is Looping??
# Repetitive/Iterative/Recursive execution of a program based on some CONDITION
#conditional looping (while loop)
#Before going with conditional looping, we need starting, ending and incremental conditions has to be defined
min_aspirant=1 #Start value
max_aspirant=10 #End Value
while min_aspirant<=max_aspirant: #Condition
    print(f"hello Aspirant Number {min_aspirant}")
    min_aspirant = min_aspirant + 1 #Incremental value
else:
    print(f"I covered all intended {max_aspirant} aspirants")
```

```python
#An equivalent program using for loop using range function?
for aspirant in list(range(1,11)): #Un-Condition
    print(f"hello Aspirant Number {aspirant}")
else:
    print(f"I covered all intended 10 aspirants")
```

```python
#Lets understand fundamentally - What is Un-Conditional Looping??
# Repetitive/Iterative/Recursive execution of a program based on some DATA/VALUES
#un-condition (for loop)
#Before going with unconditional looping we need some DATA/VALUES in a form of collections/sequence types/iterable type
values={'Meena','Vivek','Seetha','Khader'}#Seq/iter types (str/collection types)
for i in values:
    print(f"hello Aspirant: {i}")
```

```python
#Usecase2: Write a program to process a Salary increment in a loop until the salary reaches a specific target amount.
emp_name='Smarnika'
desig='SE'
sal=400000#start
avg_hike=.10#incremental
max_sal=800000#end
year=2020
#Conditinal looping (While loop)
#4 Templated clauses
#start, end, condition, increment
while sal<=max_sal:
    year+=1
    hike_amt=sal*avg_hike
    sal=sal+hike_amt
    print(f"Giving hike to {emp_name} for year {year} the hike of {hike_amt} and now the sal is {sal}")
else:
    print(f"{emp_name} reached the maximum sal of less than {max_sal}, inorder to get next hike of {sal} she have to get promoted on or before {year}")
```

```python
#Additional constructs of for & while loop
#break & continue
#Break is an additional construct only used under loop (for/while)
#Break is used to stop the iteration of the loop if some condition is met
for i in range(1,10):
    print(i)
    if i>=5:
        break#Breaking/stopping a loop execution after a condition is met
    print("inside of the loop")
print("outside of the loop")
```
