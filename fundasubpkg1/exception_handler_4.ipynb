```python
#Exception handling can be applied with notifications, mitigation, retry
def clean(waittime):
        try:#main block
                import time
                print(f"logic to do clean")
                time.sleep(100/waittime)#If i pass 0 as denominator, error occurs & program terminated , moving the control from line 4 to line
                print("Cleaning completed")
                return True
        except Exception as ex_msg:
            print(f"cleaning failed due to : {ex_msg}, some other error")#Notify
            time.sleep(10/1)#Mitigation
            return True

clean('10')
```

```python
#How to minimum write our any code with exception handler?
def fun1():
    #Without exception handler
    non_present_var1=10
    print("write loc",non_present_var1)

def fun1():
    #With exception handler
    try:#If loc inside try is success, our control will not go to except block
        non_present_var1=10
        print("write loc",non_present_var1)
    except Exception as kader:#If loc inside try is failed, except block will run
        print("some execption occured",kader)
fun1()
```

```python
#Assertion(Confidence) error
print("creating voting system for bi-election")
try:
    voter_age=int(input("enter age"))
    assert voter_age>=18#If this assertion failed, then raise an exception
    print("Allowing the voter to vote")
except AssertionError as ex_msg:
    print("AssertionError: Age of the voter should be >=18")
#assert
```

```python
#Not very important blocks, but good to know (else & finally blocks)
print("creating voting system for bi-election")
while True:
    try:#Actual logic of program, expected to run success.
        # If any issue in try block, control will go to ? except
        # If try block is succes, control will go to ? else
        voter_age=int(input("enter next voter age"))
        assert voter_age>=18#If this assertion failed, then raise an exception
    except AssertionError as ex_msg:#Mitigation logic of program, expected to run success.
        print("AssertionError: Age of the voter should be >=18")
        #continue
    else:#If try is success
        print("Allowing the voter to vote")
    finally:#If try is success or failed, finally will run at any cost
        print("bring the person out of the voting room")
        break
```
