```python
#Points to remember: Not boolean of True/False alone is considered as an operand for
#logical operators/conditional structures/looping constructs, even string, number, collections also can be used.
#False, 0, '',[],{},(), None
print(bool('None'))
```

```python
#indendation plays an important role to decide the completion of a block
#block1
if 5>5 :#1 sub block
    a=10
    b=20
    print(a+b)#block completed here due to to indendation
else:
  print("else")

#block2
print("new line of code outside of if block")
```

```python
#block1
if 5>4 :#1 sub block
    a=10
    b=20
    print(a+b)
    print("last line of if block")

#block2
if 5>3:#sb1
    print("another if")
    print(a+b)
else:#sb2 #else is optional
    print("else")

#block3
#Possible lines can be executed:
#(18 True) 19, 20, 25  - (18 False) 21 - (21 True) 22,25 - (21 False) 23 - (23 True) 24, 25 - (23 False) 25
if 5>6:#sb1
    print("if")
    print(a+b)
elif 5>5:#optional
    print("else if1")
elif 5>4:#optional
    print("else if2")
print("next block (out of above if elif elif block)")
```

```python
#block1
#(3 true) -> 4,5,8 , (3 False) -> 6, 7, 8
if 5>3:#sb1
    print("another if")
    print(a+b)
else:#sb2 #else is optional
    print("else")
print("starting of new block")
```

```python
if 5>7:#sb1
    print("if")
    print(a+b)
elif 5>6:#optional
    print("else if1")
elif 5>5:#optional
    print("else if2")
else:#default/final
    print("I will execute if none of the if, elif, elif is True")
```
