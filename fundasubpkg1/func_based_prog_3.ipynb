```python
#Irfan wrote functions
def math_table(table_you_need):
    st=1
    end=table_you_need
    while st<=end:
        print(st)
        for j in range(1,11):
            print(f"{st} x {j} = {st*j}")
        st+=1
```

```python
#Irfan kids are using this function just by passing params & getting results
#We are using these functions just by passing params & getting results
math_table(15)#parallely, distributed, concurrently...
```
