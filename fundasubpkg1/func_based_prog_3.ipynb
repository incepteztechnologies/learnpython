```python
#Anonymous Development without function
#Variable declaration
purchase=15000
pct=.1
max_amt=1000
#Actual logic
calc_off_amt=int(purchase*pct)
print(calc_off_amt)
if calc_off_amt <= max_amt:
    final_off_amt=calc_off_amt
else:
    final_off_amt=max_amt
print(final_off_amt)
#def calc_offer(purchase, pct, max_amt):
```

```python
#Converting anonymous code to function
def calc_offer(purchase, pct, max_amt):  #Variable declaration to parameters/arguments
    calc_off_amt=int(purchase*pct)
    if calc_off_amt <= max_amt:
        final_off_amt = calc_off_amt
    else:
        final_off_amt = max_amt
    return final_off_amt
```

```python
#3. Default Arguments
#create table emp(id int,name varchar(20),exp int default 0);
#insert into emp values(1,'irfan',23);
#insert into emp values(2,'sathya');
def div(x,y=1):#Default the param values if some default value has to be substituted in the runtime
#Note: Default parameters has to be after regular parameters
   z=x/y
   return z
res=div(x=10,y=20)#y is overriding 1 with 20
print(res)
res=div(10)#y is considered as default 1 (positional + default)
print(res)
res=div(x=10)
print(res)#y is considered as default 1 (named + default)
```

```python
#Converting anonymous code to function
def calc_offer(purchase,max_amt,pct=.1):  #Variable declaration to parameters/arguments
    calc_off_amt=int(purchase*pct)
    if calc_off_amt <= max_amt:
        final_off_amt = calc_off_amt
    else:
        final_off_amt = max_amt
    return final_off_amt

calc_offer(15000,1000)#Positional + Default
calc_offer(purchase=15000,max_amt=1000)#Named + Default
calc_offer(15000,1000,.1)#Positional only
calc_offer(purchase=15000,pct=.1,max_amt=1000)#Named only
```

```python
#Arbitrary Argument Functions (Dynamic/Any number of arguments) - pos/keyword
#2 collection Types
#      0    1    2 3...
tup_a=(1,'irfan',3,4) #collection/seq/complex
#b.print(tup_a[1])
#Arbitrary positional arg function
def arb_arg_fun(*emp_record):#* - tuple, ** - dict
    b.print(type(emp_record))
    b.print(emp_record)
    b.print(emp_record[0])#positional
    b.print(emp_record[1])
    b.print(emp_record[2])


arb_arg_fun('irfan',44,'chennai','velachery',600042)#emp_record=('irfan',44,'chennai')
```

```python
#fixed pos/name/default arg
def pa(a,b=0):
    return a-b
b.print(pa(10,20))#pa f
b.print(pa(a=20,b=10))#k/na f
b.print(pa(20))
```

```python
#arbitrary pos
def pa(*args):#a as a tuple
    import builtins as b
    b.print(args)
    a=args[0]
    b=args[1]
    c=args[2]
    return a-b-c
b.print(pa(10,20,30))#arbitrary
b.print(pa(10,20,30,40,50))#arbitrary
```

```python
#arbitrary named/kw

def pa(**kwargs):#kwargs as a dict
    import builtins as b
    b.print(kwargs)
    a=kwargs['a']
    b=kwargs['b']
    c=kwargs['c']
    return a-b-c
b.print(pa(a=10,b=20,c=30))#arbitrary named
b.print(pa(a=10,b=20,c=30,d=40,e=50))
```

```python
def generate_mail(fname='noname',lname='noname',domain='inceptez.com'):
    return fname+lname+'@'+domain
b.print(generate_mail("mohamed","irfan","infosys.com"))#positional (fixed)
b.print(generate_mail(domain="infosys.com",fname="mohamed",lname="irfan"))#named/keyword (fixed)
b.print(generate_mail("mohamed","irfan"))#positional+default (fixed)
b.print(generate_mail(fname="mohamed",lname="irfan"))#named/keyword + default (fixed)
b.print(generate_mail())
```

```python
def generate_mail(*args):
    fname=args[0]
    lname=args[1]
    domain=args[2]
    return fname+lname+'@'+domain
b.print(generate_mail("mohamed","irfan","inceptez.com",1,2,3))#positional (arbitrary)
```

```python
def generate_mail(**kwargs):
    fname=kwargs['fname']
    lname=kwargs['lname']
    domain=kwargs['domain']
    return fname+lname+'@'+domain
b.print(generate_mail(fname="mohamed",lname="irfan",domain="inceptez.com",misc1=1,misc2=2))#positional (arbitrary)
```

```python
#arbitrary positional
#You are building software for a weather station. Throughout the day, temperature sensors send a batch of readings back to your system. Because the number of active sensors changes depending on the location, your function must accept any number of numerical readings and calculate key summary statistics.
temp_whole_journey_10hours=30,31,35,40,42,40,32,30,20,15,10
temp_whole_journey_15hours=30,31,35,40,42,40,32,30,20,15,10,8,4,6,10,15
def temperature_stats(*args):
    import statistics
    min_temp=min(args)
    max_temp=max(args)
    avg_temp=sum(args)/len(args)
    mean_temp=statistics.mean(args)
    median_temp=statistics.median(args)
    stddev_temp=statistics.stdev(args)
    return {"min_temp":min_temp,"max_temp":max_temp,"avg_temp":avg_temp,"median_temp":median_temp,"var_temp":stddev_temp,"mean_temp":mean_temp}
b.print(temperature_stats(30,31,35,40,42,40,32,30,20,15,10))#positional + arbitrary
b.print(temperature_stats(30,31,35,40,42,40,32,30,20,15,10,8,4,6,10,15))#positional + arbitrary
```
